불합격      불합격      합격      합격
   ●-----------●----★----●-----------●
   1           2   2.5   3           4

★ = 예측할 학생

```python
from sklearn.neighbors import KNeighborsClassifier

x = [[1],[2],[3],[4]] # 공부시간
y = [0,0,1,1] # 합격 / 불합격

model = KNeighborsClassifier(n_neighbors=3) # 3,5, 7 홀수로 ... 

model.fit(x,y) 

result = model.predict([[2.5]])
print(result)
```
