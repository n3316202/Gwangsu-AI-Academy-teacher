```python
import pandas as pd

df = pd.DataFrame({
    'study_time': [1, 2, 3, 4, 5, 6],
    'pass':       [0, 0, 0, 1, 1, 1]
})

df
```

```python
from sklearn.linear_model import LogisticRegression

# x, y 분리
x = df[["study_time"]]
y = df["pass"]

print(x.shape)
print(y.shape)

# 모델 생성 
model = LogisticRegression()

# 학습
model.fit(x,y)

print(model.coef_)
print(model.intercept_)

# 예측 
pred = model.predict(x)
print(pred)


pred = model.predict([[10]]) # 10 시간일때
print(pred)

 # 10 시간일때 12 일떼
pred = model.predict([[10],[12]]) # 10 시간일때
print(pred)
```
