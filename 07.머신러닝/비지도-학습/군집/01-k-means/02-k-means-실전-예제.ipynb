# 🧠 K-Means Clustering 예제 (Scikit-learn)

## 📌 개요
- K-Means는 대표적인 비지도 학습 (Unsupervised Learning)
- 데이터를 K개의 그룹으로 자동 분류

---

## 🔥 핵심 아이디어

```text id="kmeans_core"
1. 중심점(k)을 랜덤으로 선택
2. 각 데이터 → 가장 가까운 중심에 할당
3. 중심점 재계산
4. 반복 (수렴할 때까지)

````python
# 구매횟수 / 구매금액

import pandas as pd
from sklearn.cluster import KMeans

df = pd.DataFrame({
    "횟수":[30,35,25,3,5,2],
    "금액":[100,120,90,10,15,8]
})

#print(df)

model = KMeans(n_clusters = 2)

model.fit(df)
df["cluster"] = model.labels_

print(df)
print(model.cluster_centers_)

# 분류 결과
# 0 = VIP
# 1 = 일반고객
````

````python
import pandas as pd
# 영화시청 시간과 평점
df = pd.DataFrame({
    "시간":[150,140,160,80,90,100],
    "평점":[9,10,8,3,4,2]
})

model = KMeans(n_clusters = 2)

model.fit(df)
df["cluster"] = model.labels_

print(model.cluster_centers_)
print(df)

# 시청시간이 길면 => 평점높음 => 긴 영화 선호
# 시청시간이 짧으면 => 평점 낮음 => 짧은(단편) 영화 선호
````
